## 题目1：求列表的基本统计信息

编写程序，求列表 s=[9,7,8,3,2,1,55,6] 中的元素个数、最大值、最小值、元素之和、平均值。


```python
# 方法1：使用内置函数
s = [9, 7, 8, 3, 2, 1, 55, 6]

count = len(s)  # 元素个数
max_val = max(s)  # 最大值
min_val = min(s)  # 最小值
sum_val = sum(s)  # 元素之和
avg_val = sum_val / count  # 平均值

print("方法1：使用内置函数")
print(f"列表：{s}")
print(f"元素个数：{count}")
print(f"最大值：{max_val}")
print(f"最小值：{min_val}")
print(f"元素之和：{sum_val}")
print(f"平均值：{avg_val}")
```

```python
# 方法2：使用 for i in s（直接访问元素）
s = [9, 7, 8, 3, 2, 1, 55, 6]

count = 0
max_val = s[0]
min_val = s[0]
sum_val = 0

for i in s:
    count += 1
    sum_val += i
    if i > max_val:
        max_val = i
    if i < min_val:
        min_val = i

avg_val = sum_val / count

print("方法2：使用 for i in s")
print(f"列表：{s}")
print(f"元素个数：{count}")
print(f"最大值：{max_val}")
print(f"最小值：{min_val}")
print(f"元素之和：{sum_val}")
print(f"平均值：{avg_val}")
```

```python
# 方法3：使用 for i in range(0, len(s))（通过索引访问）
s = [9, 7, 8, 3, 2, 1, 55, 6]

count = len(s)
max_val = s[0]
min_val = s[0]
sum_val = 0

for i in range(0, len(s)):
    sum_val += s[i]
    if s[i] > max_val:
        max_val = s[i]
    if s[i] < min_val:
        min_val = s[i]

avg_val = sum_val / count

print("方法3：使用 for i in range(0, len(s))")
print(f"列表：{s}")
print(f"元素个数：{count}")
print(f"最大值：{max_val}")
print(f"最小值：{min_val}")
print(f"元素之和：{sum_val}")
print(f"平均值：{avg_val}")
```

```python
# 方法4：使用 while 循环（正序访问）
s = [9, 7, 8, 3, 2, 1, 55, 6]

count = len(s)
max_val = s[0]
min_val = s[0]
sum_val = 0
i = 0

while i < len(s):
    sum_val += s[i]
    if s[i] > max_val:
        max_val = s[i]
    if s[i] < min_val:
        min_val = s[i]
    i += 1

avg_val = sum_val / count

print("方法4：使用 while 循环（正序访问）")
print(f"列表：{s}")
print(f"元素个数：{count}")
print(f"最大值：{max_val}")
print(f"最小值：{min_val}")
print(f"元素之和：{sum_val}")
print(f"平均值：{avg_val}")
```

```python
# 方法5：使用 while 循环（反序访问）
s = [9, 7, 8, 3, 2, 1, 55, 6]

count = len(s)
max_val = s[0]
min_val = s[0]
sum_val = 0
i = len(s) - 1

while i >= 0:
    sum_val += s[i]
    if s[i] > max_val:
        max_val = s[i]
    if s[i] < min_val:
        min_val = s[i]
    i -= 1

avg_val = sum_val / count

print("方法5：使用 while 循环（反序访问）")
print(f"列表：{s}")
print(f"元素个数：{count}")
print(f"最大值：{max_val}")
print(f"最小值：{min_val}")
print(f"元素之和：{sum_val}")
print(f"平均值：{avg_val}")
```

## 题目2：将列表中的偶数变成它的平方

编写程序，将列表 s=[9, 7, 8, 3, 2, 1, 5, 6] 中的偶数变成它的平方，奇数保持不变。


```python
# 方法1：使用 for 循环直接修改列表
s = [9, 7, 8, 3, 2, 1, 5, 6]
print(f"原列表：{s}")

for i in range(len(s)):
    if (s[i] % 2) == 0:  # 判断是否为偶数
        s[i] = s[i] ** 2  # 将偶数变成它的平方

print(f"修改后：{s}")
```

```python
# 方法2：使用 while 循环
s = [9, 7, 8, 3, 2, 1, 5, 6]
print(f"原列表：{s}")

i = 0
while i < len(s):
    if (s[i] % 2) == 0:
        s[i] = s[i] ** 2
    i += 1

print(f"修改后：{s}")
```

```python
# 方法3：使用列表推导式（更加 Pythonic）
s = [9, 7, 8, 3, 2, 1, 5, 6]
print(f"原列表：{s}")

s = [x**2 if x % 2 == 0 else x for x in s]

print(f"修改后：{s}")
```

## 题目3：输入字符串，输出每个字符的ASCII码列表

编写程序，输入字符串，为其每个字符的 ASCII 码形成列表并输出。


```python
# 方法1：使用 append 方法（通过索引访问）
string = input("请输入一个字符串：")
ascii_list = []  # 创建空列表

for i in range(len(string)):
    ascii_code = ord(string[i])  # 获取字符的 ASCII 码（Unicode 码）
    ascii_list.append(ascii_code)  # 追加到列表尾部

print(f"输入的字符串：{string}")
print(f"ASCII 码列表：{ascii_list}")
```

```python
# 方法2：直接遍历字符串
string = input("请输入一个字符串：")
ascii_list = []

for char in string:
    ascii_list.append(ord(char))

print(f"输入的字符串：{string}")
print(f"ASCII 码列表：{ascii_list}")
```

```python
# 方法3：使用列表推导式
string = input("请输入一个字符串：")
ascii_list = [ord(char) for char in string]

print(f"输入的字符串：{string}")
print(f"ASCII 码列表：{ascii_list}")
```
