## 3.1 重量计算

月球上物体的重量是在地球上的16.5%，假如你在地球上每年增长0.5 kg，编写程序输出未来10年你在地球和月球上的体重状况。


```python
# 3.1 重量计算程序

current_weight = float(input("请输入您当前的体重(kg): "))

moon_weight_ratio = 0.165

yearly_gain = 0.5

print("\n未来10年体重状况:")
print("年份\t地球体重(kg)\t月球体重(kg)")

for year in range(1, 11):
    earth_weight = current_weight + (year * yearly_gain)
    moon_weight = earth_weight * moon_weight_ratio
    
    print(f"{year}\t{earth_weight:.1f}\t\t{moon_weight:.2f}")

print(f"10年后地球体重: {current_weight + 10 * yearly_gain:.1f} kg")
print(f"10年后月球体重: {(current_weight + 10 * yearly_gain) * moon_weight_ratio:.2f} kg")
```

## 3.2 天天向上

尽管每天坚持，但人的能力发展并不是无限的，它符合特定模型。假设能力增长符合如下带有平台期的模型：以7天为周期，连续学习3天能力值不变，从第4天开始至第7天每天能力增长为前一天的1%。如果7天中有1天间断学习，则周期从头计算。请编写程序回答，如果初始能力值为1，连续学习365天后能力值是多少？


```python
# 3.2 天天向上能力增长模型

def calculate_ability_growth(days):
    ability = 1.0  
    cycle_day = 0  
    
    print("学习进度和能力变化:")
    print("天数\t周期内天数\t能力值")
    
    for day in range(1, days + 1):
        cycle_day = (day - 1) % 7  
        
        if cycle_day < 3:
            pass  
        else:
            ability *= 1.01
        
        if day % 10 == 0 or day <= 10:
            print(f"{day}\t{cycle_day + 1}\t\t{ability:.6f}")
    
    return ability

total_days = 365
final_ability = calculate_ability_growth(total_days)

print(f"连续学习{total_days}天后，能力值为: {final_ability:.6f}")
print(f"相比初始能力值增长了: {(final_ability - 1) * 100:.2f}%")
```

## 3.4 回文数判断

设n是一个任意自然数，如果n的各位数字反向排列所得自然数与n相等，则n被称为回文数。从键盘输入一个5位数字，请编写程序判断这个数字是不是回文数。


```python
# 3.4 回文数判断程序

def check_palindrome(number):
    num_str = str(number)
    
    if len(num_str) != 5:
        return "请输入5位数字"
    
    if num_str == num_str[::-1]:
        return f"{number} 是回文数"
    else:
        return f"{number} 不是回文数"

try:
    user_input = input("请输入一个5位数字: ")
    number = int(user_input)
    
    result = check_palindrome(number)
    print(result)
    
except ValueError:
    print("错误：请输入有效的数字！")
except Exception as e:
    print(f"发生错误: {e}")
```
