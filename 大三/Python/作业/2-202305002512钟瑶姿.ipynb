## 2.1 温度转换程序的修改

```python
Temp = eval(input("请输入温度数值: "))  
Type = input("请输入温度单位（C/F）: ") 

if Type in ['F', 'f']:
    C = (Temp - 32) / 1.8
    print("转换后的温度是{:.0f}C".format(C)) 
elif Type in ['C', 'c']:
    F = 1.8 * Temp + 32
    print("转换后的温度是{:.0f}F".format(F))  
else:
    print("输入格式错误")
```

## 2.2 汇率兑换程序

```python
MoneyStr = input("请输入带有符号的金额值:")
if MoneyStr[-1] in ['D','d']:
    RMB = eval(MoneyStr[0:-1]) * 6
    print("转换后的人民币是{:.2f}元".format(RMB))
elif MoneyStr[-1] in ['R','r']:
    USD = eval(MoneyStr[0:-1]) / 6
    print("转换后的美元是{:.2f}元".format(USD))
else:
    print("输入格式错误")
```

## 2.3 重量转换程序

```python
WeightStr = input("请输入带有符号的重量值:")
if WeightStr[-1] in ['K','k']:
    Pound = eval(WeightStr[0:-1]) * 2.2046
    print("转换后的磅数是{:.2f}磅".format(Pound))
elif WeightStr[-1] in ['P','p']:
    Kg = eval(WeightStr[0:-1]) / 2.2046
    print("转换后的千克数是{:.2f}千克".format(Kg))
else:
    print("输入格式错误")
```
