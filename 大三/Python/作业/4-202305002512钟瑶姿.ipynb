## 4.2 不同字符统计
从键盘输入一行字符串，统计并输出其中英文字符、数字、空格和其他字符的个数。

```python
input_line = input("请输入一行字符: ")

english_chars = 0
digits = 0
spaces = 0
others = 0

for char in input_line:
    if 'a' <= char.lower() <= 'z':
        english_chars += 1
    elif char.isdigit():
        digits += 1
    elif char == ' ':
        spaces += 1
    else:
        others += 1

print(f"英文字符个数: {english_chars}")
print(f"数字个数: {digits}")
print(f"空格个数: {spaces}")
print(f"其他字符个数: {others}")
```

## 4.3 最大公约数与最小公倍数
从键盘读入两个正整数，编写程序求出这两个整数的最大公约数和最小公倍数。

```python
from math import gcd

def lcm(a: int, b: int) -> int:
    if a == 0 or b == 0:
        return 0
    return abs(a // gcd(a, b) * b)

a = int(input("请输入第一个正整数："))
b = int(input("请输入第二个正整数："))

result_gcd = gcd(a, b)
result_lcm = lcm(a, b)
print(f"最大公约数: {result_gcd}")
print(f"最小公倍数: {result_lcm}")
```

## 4.4 猜数字游戏（范围0~1000）
计算机随机生成一个整数，用户反复猜测，程序根据大小给出提示，直到猜中为止。

```python
import random

print("我已经想好了一个0到1000之间的数字，请开始猜测吧！")

target = random.randint(0, 1000)
attempts = 0

while True:
    guess = int(input("请输入0~1000之间的整数："))
    attempts += 1
    
    if guess < 0 or guess > 1000:
        print("超出范围，请重试！")
        continue
    elif guess < target:
        print(f"{guess}小了！")
    elif guess > target:
        print(f"{guess}大了！")
    else:
        print(f"恭喜猜中{guess}！共用{attempts}次。")
        break
```

## 4.6 闰年判断
编写程序：对于用户输入的年份，判断是否为闰年。规则：能被4整除且不能被100整除，或能被400整除的年份为闰年。

```python
def is_leap_year(year: int) -> bool:
    return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)

year = int(input("请输入年份："))
result = "闰年" if is_leap_year(year) else "平年"
print(f"{year}年是{result}")
```
