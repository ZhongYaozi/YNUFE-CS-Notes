```python
num = input("请输入一个 5 位数字：")
def check_palindrome(num):
    num_str=str(num)
    if len(num_str) != 5:
        return "请输入5位数字"
    if num_str == num_str[::-1]:
        return f"{num}是回文数"
    else:
        print("{} 不是回文数".format(num))
result = check_palindrome(num)
print(result)
```
